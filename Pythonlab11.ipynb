```python
def calculate_fare(ticket_price, passengers):
   total_amount = ticket_price * passengers
   return total_amount
```

```python
final_fare = calculate_fare(2450, 2)
print("Final Fare :", final_fare)
```

```python
def verify_passenger(age):
   if age >= 18:
       return "Eligible for Reservation"
   else:
       return "Not Eligible"
```

```python
result = verify_passenger(25)
print(result)
```

```python
def fact(n):
    if n == 0 or n == 1:
         return 1
    return n * fact(n - 1)

num = int(input("Enter a number: "))
print("Factorial:",fact(num))
```

```python
def fact(n):
  if n == 0 or n == 1:
    return 1
  elif n > 1:
    return n * fact(n-1)
  else:
    print("Factorial of only positive integer is allow")
num = int(input("Enter a number:"))
print("Factorial:",fact(num))
```
